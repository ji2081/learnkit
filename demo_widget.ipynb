# 위젯 데모 — learnkit `Lesson`을 주피터에서

드롭다운(갈래)·토글(난이도)·입력(내 데이터)을 바꾸면 결과가 즉시 갱신됩니다.
'바꾸기'엔 힌트, '만들기'엔 코드와 도전 과제가 함께 나옵니다.

> 설치: `pip install ipywidgets rich`

```python
from learnkit import Lesson

def 번호매기기(items):
    return [f"{i+1}. {x}" for i, x in enumerate(items)]

lesson = Lesson(
    title="반복으로 '내 것' 정리하기",
    build=번호매기기,
    branches={"인문": ["좋아하는 책", "영화", "노래"], "숫자": ["용돈 1000", "간식 2500"]},
    hint="branches의 값을 네가 좋아하는 것들로 바꿔봐.",
    challenge="번호 대신 '⭐ 항목' 형태로 바꿔보자.",
    big=True, high_contrast=True,
)

lesson.as_widget()
```
